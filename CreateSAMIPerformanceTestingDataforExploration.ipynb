```python
import os
import json
import shutil
import random
from datetime import datetime, timezone
payload_folder = "/content/pytest_ai4l_scripts/SAMI/fixtures/v1p2_2025_08_25_133949"
#postman_folder='/content/pytest_ai4l_scripts/performance_testing/postman/'
archive_base_folder='/content/pytest_ai4l_scripts/archive'
```

```python
min_events_per_envelope=5
max_events_per_envelope=35
```

```python

payload_performance_testing=os.path.join(payload_folder,'performance_testing')
shutil.rmtree(payload_performance_testing,ignore_errors=True)
os.makedirs(payload_performance_testing,exist_ok=True)
```

```python
import os
import json

# Assuming payload_folder and payload_performance_testing are defined
# somewhere before this code block. For example:
# payload_folder = 'path/to/your/payloads'
# payload_performance_testing = 'path/to/your/performance_test_payloads'
# min_events_per_envelope = 1
# max_events_per_envelope = 20

for sub_dir in os.listdir(payload_folder):
  if 'performance_testing' not in sub_dir:

    for i in range(min_events_per_envelope, max_events_per_envelope, 5):
        # Initialize a list to hold all the data elements
        all_data = []

        # Flag to check if we have enough data
        data_collected = False

        # Iterate through all files
        for root, dirs, files in os.walk(os.path.join(payload_folder,sub_dir)):
            for file in files:

                # Check for JSON files and exclude the performance testing folder
                if file.endswith('.json'):

                    # Check if we have already collected enough data for this 'i'
                    if len(all_data) >= i:
                        data_collected = True
                        break  # Exit the file loop

                    with open(os.path.join(root, file)) as f:
                        payload = json.load(f)
                        file_data = payload['data']

                        # Determine how many elements to take from the current file
                        elements_to_take = min(len(file_data), i - len(all_data))

                        # Add the elements to our collection
                        all_data.extend(file_data[:elements_to_take])

            if data_collected:
                break # Exit the root/dirs loop

        # If we successfully collected the required number of elements
        if len(all_data) == i:
            # Create the new payload with the combined data
            new_payload = {
                'sensor': payload['sensor'],  # Assumes all files have the same metadata. You may need to adjust this.
                'sendTime': payload['sendTime'],
                'dataVersion': payload['dataVersion'],
                'data': all_data
            }

            # Write the combined payload to a new file
            print('len of all data in the data array ',len(all_data))
            with open(os.path.join(payload_performance_testing, f'{root.split('/')[-1]}_{len(all_data)}.json'), 'w') as f:
                json.dump(new_payload, f, indent=2)
        else:
            print(f"Could not find enough data to create a combined payload of size {i}.")
```
