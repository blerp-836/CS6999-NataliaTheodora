```python
import random
import string

#create random string generator
def generate_random_alphanumeric_string(length):
    """
    Generates a random string consisting of uppercase letters, lowercase letters, and digits.

    Args:
        length (int): The desired length of the random string.

    Returns:
        str: A randomly generated alphanumeric string.
    """
    # Define the set of characters to choose from:
    # string.ascii_letters includes 'abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ'
    # string.digits includes '0123456789'
    characters = string.ascii_letters + string.digits

    # Use random.choice to pick a character from the 'characters' set for 'length' times
    # Then, join them together to form the final string
    random_string = ''.join(random.choice(characters) for i in range(length))

    return random_string

```

```python
#generate formatted date
from datetime import datetime
import pytz

def get_current_time(return_type='utc'):
  # Get the current UTC time
  utc_now = datetime.now(pytz.utc)

  # Convert to Eastern Daylight Time (EDT)
  edt_timezone = pytz.timezone('America/New_York')
  edt_now = utc_now.astimezone(edt_timezone)

  # Format the datetime object as requested
  # Example: "June 30, 2025, 6:11:40 PM Eastern Daylight Time"
  formatted_time = edt_now.strftime("%B %d, %Y, %I:%M:%S %p %Z")
  if return_type=='edt':
    return edt_now.isoformat()
  elif return_type=='strf':
    return formatted_time
  elif return_type=='utc':
    return utc_now.isoformat()
print(get_current_time('edt'))
```

```python
import json
import copy
def delete_key_and_save_to_json(original_dict, keys_to_delete, output_filename="output.json"):
    """
    Creates a new dictionary from an original one, deleting specified keys from a list,
    and then saves the new dictionary to a JSON file.

    Args:
        original_dict (dict): The dictionary to start with. This dictionary
                              will not be modified.
        keys_to_delete (list): A list of strings, where each string is the name
                               of a key to be deleted.
        output_filename (str): The name of the JSON file to save the new
                               dictionary to. Defaults to "output.json".
    """
    # Create a new dictionary to store the modified data

    #create a copy of the original dictionary
    new_dict=copy.deepcopy(original_dict)


    # Iterate through the items of the original dictionary
    for key_to_delete in keys_to_delete:
      print('key to delete ',key_to_delete)
      if '.' in key_to_delete:
        parent_key=key_to_delete.split('.')[0]
        child_key=key_to_delete.split('.')[1]
        if parent_key in new_dict:
          #check if key exists
          if child_key in new_dict[parent_key]:
            del new_dict[parent_key][child_key]
      else:
        if key_to_delete in new_dict:
          if key_to_delete in new_dict:
            del new_dict[key_to_delete]



        # If it IS in the keys_to_delete list, we simply do nothing, effectively "deleting" it
        # from the new_dict

    # Print both dictionaries to show the original is preserved
    # print("Original Dictionary:")
    # print(original_dict)
    # print("\nNew Dictionary with Keys Deleted:")
    # print(new_dict)

    # Write the new dictionary to a JSON file
    try:
        with open(output_filename, 'w', encoding='utf-8') as f:
            # Use json.dump() to write the dictionary to the file
            # indent=4 makes the JSON output human-readable with indentation
            json.dump(new_dict, f, indent=4)
        print(f"\nSuccessfully wrote the new dictionary to '{output_filename}'")
    except IOError as e:
        print(f"Error writing to file '{output_filename}': {e}")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")


```

```python
import json
import copy

def update_key_value_and_save_to_json(original_dict, updates_mapping, output_filename="output.json"):
    """
    Takes an original dictionary, creates a new dictionary, updates the values
    for multiple specified keys based on a mapping, and then saves the new dictionary
    to a JSON file. The original dictionary passed to the function remains untouched.

    Args:
        original_dict (dict): The dictionary to start with. This dictionary
                              will not be modified.
        updates_mapping (dict): A dictionary where keys are the names of the keys
                                to be updated, and values are their new values.
                                Example: {"key1": "newValue1", "key2": 123}
        output_filename (str): The name of the JSON file to save the new
                               dictionary to. Defaults to "output.json".
    """
    # Create a new dictionary by copying the original to preserve the original
    new_dict = copy.deepcopy(original_dict)

    # Iterate through the updates_mapping and apply each update to the new dictionary
    for key_to_update, new_value in updates_mapping.items():
        # This will update the value if the key exists, or add it if it doesn't
        if '.' in key_to_update:
          key_to_update_parent=key_to_update.split('.')[0]
          key_to_update_child=key_to_update.split('.')[1]
          new_dict[key_to_update_parent][key_to_update_child] = new_value
        else:
          new_dict[key_to_update] = new_value

    # Print both dictionaries to show the original is preserved
    # print("\nOriginal Dictionary (passed to function):")
    # print(original_dict)
    # print("\nNew Dictionary with Updated Values:")
    # print(new_dict)

    # Write the new dictionary to a JSON file
    try:
        with open(output_filename, 'w', encoding='utf-8') as f:
            # Use json.dump() to write the dictionary to the file
            # indent=4 makes the JSON output human-readable with indentation
            json.dump(new_dict, f, indent=4)
        print(f"\nSuccessfully wrote the new dictionary to '{output_filename}'")
    except IOError as e:
        print(f"Error writing to file '{output_filename}': {e}")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
```

# Create file for postman runner


```python
from datetime import datetime, timezone
def create_header(events,sensor_id=1):
    utc_now=datetime.now(timezone.utc).isoformat(timespec='milliseconds')
    #events is a list object. it can contain one or more dictionary object.
    return {
        "sensor": f"https://example.edu/sensors/{sensor_id}",
        "sendTime": utc_now.split('+')[0]+'Z',
        "dataVersion": "http://purl.imsglobal.org/ctx/caliper/v1p2",
        "data": events
    }
```

```python
import json
#create directory if not exist
os.makedirs('./SAMI/MessageEventSchemaTest/Payload', exist_ok=True)
for filename in os.listdir('./SAMI/MessageEventSchemaTest'):
  if 'Event' in filename and 'Payload' not in filename and filename.endswith('.json') and 'with' in filename:
    print('filename ',filename)
    with open(os.path.join('./SAMI/MessageEventSchemaTest',filename), 'r', encoding='utf-8') as f:
      event = json.load(f)
    # # payload takes in list object as events
    payload = create_header([event],sensor_id=10)
    with open(os.path.join('./SAMI/MessageEventSchemaTest/Payload',filename), 'w') as f:
      json.dump(payload, f, indent=4)
```

```python

payload_folder = './SAMI/MessageEventSchemaTest/Payload'
os.makedirs(os.path.join(payload_folder,'postman'), exist_ok=True)
#timestamp the file with current date
timestamp = datetime.now().strftime("%Y_%m_%d_%H%M%S")
postman_filename = f"CaliperEventMessagePosted_{timestamp}.json"
#do not traverse the child folder
with open(os.path.join(payload_folder,f'postman/{postman_filename}'), "w") as outfile:
    json.dump(
        [{"body": json.load(open(f"{payload_folder}/{filename}"))}
         for filename in sorted(os.listdir(payload_folder)) if 'Event' in filename and 'Envelope' not in filename and filename.endswith('.json')],
        outfile, indent=2
    )
```
